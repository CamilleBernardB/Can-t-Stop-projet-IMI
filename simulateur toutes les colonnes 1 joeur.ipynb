```python
def simulation_complete_1player(lancers, Ns, choose_action, verbose=False):
    progress = {s: 0 for s in range(2, 13)}
    temp_progress = {}
    n_turns = 0
    i = 0
    
    while sum(progress[s] == Ns[s] for s in range(2, 13)) < 3:
        temp_progress = {}
        turn_over = False
        
        while not turn_over:
            d1, d2, d3, d4 = lancers[i]
            i += 1
            
            pairings = [
                tuple(sorted((d1 + d2, d3 + d4))),
                tuple(sorted((d1 + d3, d2 + d4))),
                tuple(sorted((d1 + d4, d2 + d3))),
            ]
            
            legal_actions = []
            for a in pairings:
                if action_is_legal(a, progress, temp_progress, Ns):
                    legal_actions.append(a)
            
            if not legal_actions:
                temp_progress = {}
                turn_over = True
                if verbose:
                    print("Bust")
                continue
            
            action, stop_now = choose_action(progress, temp_progress, legal_actions)
            temp_progress = apply_action(action, progress, temp_progress, Ns)
            
            if verbose:
                print("Roll:", (d1, d2, d3, d4))
                print("Legal:", legal_actions)
                print("Chosen:", action)
                print("Temp:", temp_progress)
            
            if stop_now:
                for s, inc in temp_progress.items():
                    progress[s] = min(progress[s] + inc, Ns[s])
                temp_progress = {}
                turn_over = True
        
        n_turns += 1
    
    return n_turns, progress
```
